### Task - 7
**Your Assignment:** Draw a horizontal bar chart of attrition\_pct, with the highest-attrition department at the top. Colour only that bar red and all other bars grey, and make the bars 0.6 thick. Print each value at the end of its bar with one decimal place followed by a percent sign (for example 18.2%), with a gap of 4 points. Make the horizontal axis visible from 0 to 22. Add dotted vertical grid lines only. Label the axes ("Attrition Rate (%)", "Department") and set the title "Employee Attrition by Department". Display the chart.

```python
# Sample Data
# ordered from lowest to highest attrition
Departments = ["Legal", "Finance", "HR", "Marketing", "Operations", "Sales", "Support"]
Attrition_pct = [4.2, 5.8, 7.1, 9.4, 11.3, 14.6, 18.2]
```

```python
Colours = []
for Attrition in Attrition_pct:
    if (Attrition == max(Attrition_pct)):
        Colours.append("red")
    else:
        Colours.append("grey")
        

Bar_4 = plt.barh(Departments , Attrition_pct , color = Colours , height = 0.6)

plt.bar_label(Bar_4 , fmt="%.1f%%" , padding = 4)
plt.xlim(0,22)
plt.grid(linestyle = ":" , axis = "x")
plt.xlabel("Attrition Rate (%)")
plt.ylabel("Department")
plt.title("Employee Attrition by Department")

plt.tight_layout()
plt.show()
```

Label the axes ("Region", "Profit / Loss (USD thousands)") and set the title "Regional Profit and Loss: FY2026". Display the chart.

```python
# Sample Data
Regions = ["North", "South", "East", "West", "Central", "Islands"]
Profit_k = [145, -32, 210, 88, -57, 14]   # thousands of USD; negative = loss
```

```python
plt.figure(figsize=(7,7))

colors = []

for value in Profit_k:
    if (value < 0):
        colors.append("red")
    else:
        colors.append("green")

Bar_11 = plt.bar(Regions,
                 Profit_k,
                 width=0.6,
                 edgecolor = "black",
                 color = colors
                )
plt.bar_label(Bar_11,
              Profit_k,
              fontsize = 10,
              padding = 3
             )
plt.yticks(range(-100 , 250 , 50))
plt.grid(axis = "y",
         linestyle = ":"
        )
plt.xlabel("Region")
plt.ylabel("Profit / Loss (USD thousands)")
plt.title("Regional Profit and Loss: FY2026")

plt.show()
```
