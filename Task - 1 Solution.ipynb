```python
# Sample Data
Agents = ["Anita", "Rahul", "Meera", "Sanjay", "Priya"]
Tickets_resolved = [132, 118, 145, 97, 124]
```

## Task - 17
Draw a bar chart of tickets_resolved by agent, a width of 0.6, black outlines of thickness 1 and 85% opacity. Print each bar's value
above it in bold, size 11, with a gap of 2 points. Show the agent names rotated 30 degrees.
Make the vertical axis visible from 0 to 170 and show a value every 20, from 0 to 160. Add dotted lines running horizontally only. Name the data "Tickets Resolved" in a legend. Label
the axes ("Agent", "Tickets Resolved") and set the title "Monthly Resolution Leaderboard".
Display the chart.

```python

Tickets_Bar = plt.bar(
    Agents,
    Tickets_resolved,
    color="white",
    edgecolor="black",
    linewidth=1,
    alpha=0.85,
    label="Ticket Resolved"
)

plt.bar_label(
    Tickets_Bar,
    labels=Tickets_resolved,
    fontsize=11,
    padding=2,
    fontweight="bold"
)

plt.xticks(rotation=30)

plt.ylim(0, 170)
plt.yticks(range(0, 171, 20))

plt.xlabel("Agent")
plt.ylabel("Tickets Resolved")
plt.title("Monthly Resolution Leaderboard")

plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```
